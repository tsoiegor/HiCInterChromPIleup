**LOAD LIBRARIES**

```python
from HiCInterChromPIleup import RablPileup
import cooler
import os
```

**SPECIFY PATHS TO FILES**

```python
basedir = '/storage3/estsoi/172.26.5.30/ftp/by_User/estsoi/degron_archive/100k_cool'
cool_files_rad_aux = [cooler.Cooler(os.path.join(basedir, file + '::/resolutions/100000')) for file in ['RA1_10_kb_100k.mcool', 'RA2_10_kb_100k.mcool']]
cool_files_rad_control = [cooler.Cooler(os.path.join(basedir, file + '::/resolutions/100000')) for file in ['RC1_10_kb_100k.mcool', 'RC2_10_kb_100k.mcool']]

cool_files_mcph1_aux = [cooler.Cooler(os.path.join(basedir, file + '::/resolutions/100000')) for file in ['dMR70-A_10_kb_100k.mcool', 'dMR231-A_10_kb_100k.mcool']]
cool_files_mcph1_control = [cooler.Cooler(os.path.join(basedir, file + '::/resolutions/100000')) for file in ['dMR70-C_10_kb_100k.mcool', 'dMR231-C_10_kb_100k.mcool']]

cool_files_caph2_aux = [cooler.Cooler(os.path.join(basedir, file + '::/resolutions/100000')) for file in ['CapH2_aux_r1_10_kb_100k.mcool', 'CapH2_aux_r2_10_kb_100k.mcool']]
cool_files_caph2_control = [cooler.Cooler(os.path.join(basedir, file + '::/resolutions/100000')) for file in ['CapH2_C_r1_10_kb_100k.mcool', 'CapH2_C_r2_10_kb_100k.mcool']]
```

**CREATE RablPileup OBJECTS: THEY PERFORM PILEUP IN \_\_INIT\_\_ AND STORE RESULT** 

```python
control_rad = RablPileup(coolers=cool_files_rad_control, skip_top=2, skip_bottom=1, dim=53)
aux_rad = RablPileup(coolers=cool_files_rad_aux, skip_top=2, skip_bottom=1, dim=53)

aux_mcph1 = RablPileup(coolers=cool_files_mcph1_aux, skip_top=2, skip_bottom=1, dim=53)
control_mcph1 = RablPileup(coolers=cool_files_mcph1_control, skip_top=2, skip_bottom=1, dim=53)

aux_caph2 = RablPileup(coolers=cool_files_caph2_aux, skip_top=2, skip_bottom=1, dim=53)
control_caph2 = RablPileup(coolers=cool_files_caph2_control, skip_top=2, skip_bottom=1, dim=53)
```

**GENERATING PLOTS**

.plot() method has arguments for saving result:\
***plot_name*** : required for saving, should not include extension\
***save***=False \
***cmap***='coolwarm' : not applicable for difference plot, would be reset to 'viridis' automatically\
***title***=True \
***header_dict***={'title': 'res: 100_000, size: 50', 'fontsize': 13} \
***dpi***=600 \
***vmin***=None \
***vmax***=None \
***log2***=False : applies log2() to matrix values\
***percentiles***=False : if True, would generate matrix of percentiles\
***folder_to_save***="./" 

```python
control_rad.plot()
```

**RablPileup class supports subtraction: it should be used for plotting difference**

```python
diff_caph2 = aux_caph2 - control_caph2
diff_caph2.plot(vmax=0.00015, vmin=-0.00015)
```

```python
diff_mcph1_rad = control_mcph1 - control_rad
diff_mcph1_rad.plot(vmax=0.00015, vmin=-0.00015)
```

```python
diff_mcph1 = aux_mcph1 - control_mcph1
diff_mcph1.plot(vmax=0.00015, vmin=-0.00015)
```
